```python
import numpy as np
import pandas as pd
import joblib
```

```python
model = joblib.load("budgetease_multi_model (2).pkl")
scaler = joblib.load("scaler (1).pkl")
le_spender = joblib.load("spender_encoder (1).pkl")
le_income = joblib.load("income_encoder (1).pkl")

print("✅ Model and preprocessors loaded")
```

```python
age_map = {
    1: 0,   # 18–20
    2: 1,   # 21–23
    3: 2,   # 24–25
    4: 3    # Above 25
}
```

```python
income_source_map = {
    1: 0,   # Family support
    2: 1,   # Mahapola / Bursary
    3: 2,   # Part-time job
    4: 3    # Other
}
```

```python
expense_map = {
    1: 0,   # Essentials
    2: 1,   # Academic
    3: 2,   # Leisure
    4: 3    # Other
}
```

```python
tracking_map = {
    1: 0,   # Yes regularly
    2: 1,   # Sometimes
    3: 2,   # Rarely
    4: 3    # Never
}
```

```python
amount_map = {
    1: 8000,
    2: 15000,
    3: 25000,
    4: 35000
}
```

```python
def manual_test_prediction():
    print("\n🧪 MANUAL USER TESTING\n")

    # Age
    print("Age: 1.18–20  2.21–23  3.24–25  4.Above 25")
    age = age_map[int(input("Enter choice: "))]

    # Income sources
    print("\nIncome sources (1.Family 2.Mahapola 3.Part-time 4.Other)")
    income_choices = list(map(int, input("Enter comma-separated values: ").split(",")))
    income_source = np.mean([income_source_map[i] for i in income_choices])

    # Average income
    print("\nAverage monthly income(1.Less than 10000 2.10000-20000 3.20000-30000 4.Above 30000)")
    avg_income = amount_map[int(input("Enter choice : "))]

    # Expense categories
    print("\nMain expense categories (1.Essentials 2.Academic 3.Leisure 4.Other)")
    expense_choices = list(map(int, input("Enter comma-separated values: ").split(',')))
    main_expense = np.mean([expense_map[i] for i in expense_choices])

    # Expense tracking habit
    print("\nExpense tracking habit(1.Yes regularly 2.Sometimes 3.Rarely 4.Never)")
    tracking = tracking_map[int(input("Enter choice : "))]

    # Monthly spending
    print("\nApprox monthly spending(1.Less than 10000 2.10000-20000 3.20000-30000 4.Above 30000)")
    approx_spend = amount_map[int(input("Enter choice : "))]

    # Spending percentages
    essentials_pct = float(input("Essentials %: "))
    academic_pct = float(input("Academic %: "))
    leisure_pct = float(input("Leisure %: "))
    other_pct = float(input("Other %: "))

    input_df = pd.DataFrame([[
        age,
        main_expense,
        tracking,
        approx_spend,
        avg_income,
        essentials_pct,
        academic_pct,
        leisure_pct,
        other_pct,
        income_source
    ]], columns=[
        "Age",
        "Main_Expense_Categories",
        "Do you usually track your expenses?",
        "Approx_Monthly_Spending",
        "Avg_Monthly_Income",
        "Essentials_pct",
        "Academic_pct",
        "Leisure_pct",
        "Other_pct",
        "Main_Income_Source"
    ])

    print("\nGenerated DataFrame:")
    print(input_df)

    input_scaled = scaler.transform(input_df)
    print("\nScaled Input:")
    print(input_scaled)

    prediction = model.predict(input_scaled)
    spender_type = le_spender.inverse_transform([prediction[0][0]])[0]
    income_type = le_income.inverse_transform([prediction[0][1]])[0]

    print(f"\nPredicted Spender Type: {spender_type}")
    print(f"Predicted Income Type: {income_type}")

    print("\n🎯 PREDICTION RESULT")
    print("Spender Type :", spender_type)
    print("Income Type  :", income_type)


    # Further processing (encoding, prediction) would go here
```

```python
manual_test_prediction()
```
